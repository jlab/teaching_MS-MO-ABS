The huge advantage of dictionaries over lists is, that we do not have to search to the whole thing if we are looking for something specific. Lets say we have a database overy every student of the university. If we, instead, used a dictionary, where the keys are the names, we could call the information for each student by name in one step. 

For larger datasets - which we commonly encounter in bioinformatics - this can make quite a difference in runtime.

### Integers vs Floating-Point Numbers

Generally, computers can store numbers in two different ways: Integers ("ganze Zahlen"; e.g. -3234, -4, 0, 345) are stored directly as their binary representation, with an additional sign (+/-). This cannot hold any non-integer decimal numbers (e.g. -123.734, 0.3, 23.2343). Alternatively, numbers can be stored as floating-point numbers (float):
$-123734 * 10^{⁻3}$

Here, the differnt components (sign (-), exponent (3), and significand (123734)) are stored. If you want a number that is technically an integer to be stored as a float, you can to write it with a `.` or `.0` at the end (`10` -> `10.0`). Fortunately for you, Python mostly decides if a number has to be stored as an intger or a float on its own. Still there may be some situations, where it's relevant for you to know the difference. 

For example, divisions and multiplications by floats always return a float:

````python
print(34 * 12.0)
print(type(34 * 12.0))
print(34/2) 
print(type(34/2)) # returns float type, even though it is a whole-number relsult
````

This can cause issues in situations where an integer is required. For example indices always have to be integers:

````python
my_list = [1, 2, 4]
print(my_list[1]) # works
print(my_list[1.0]) # causes error -> error message tells you what's wrong
````

## If-Else Statements

We now want to use this logic to take conditional actions - we want to tell the computer: If this condition applies, do that; if another condition applies, do something else, and if none of the conditions apply, do another thing. We can do this with if-else statements:

````python
# modify the conditons and check what happens
food_at_home = False
time_to_cook = True

if food_at_home:
    # if the first condition applies, we enter this code block
    print("I'll go home and eat.")
elif time_to_cook: # "elif" is short for "else if"
    # the first condition does not apply, but this one does
    print("I'll go grocery shopping first, and then go home and cook.")
else:
    # none of the conditions above applied
    print("I'll grab some take-out.")
````

If the listed condition applies, the code block below is executed. In Python, code blocks are separated by tabs.

````python
condition = False # try changing the contidion to True

if condition:
    print("condition is true")
    print("this is inside the code block and is executed if the condition applies")
print("this is outside (after) the code block and is executed in any case")
````

If there is no else-block, nothing happens if the conditon does not apply.

````python
# try changing the conditions
condition_1 = False
condition_2 = False

if condition_1:
    print("contidion 1 is true")
if condition_2: # what would be different if this said "elif"?
    print("condition 2 is true")
````

## 4. Loops

Using a script to do something instead of doing it manually is especially useful if you have to do the same thing, or a similar thing over and over again, because you can use loops: a piece of code that is repeatedly executed, either a certain number of times or until a condition is met. 

Python has two kinds of loops: 
- `for`-loops: They require a list or series of some sort, over which it iterates. The first execution of the code block looks at the first element, the second at the second element, and so on. If all elements have been looked at, it stops.
- `while`-loops: The code block is repeated while a condition applies. If the condtition changes, the loop stops.

### `for`-Loops

The syntax for `for`-loops is as follows:

```py
for element in a_list:
    # do something
```

`a_list` is the variable name of the list you want to iterate over. `element` is the name you will need to refer to each element as - you can pick this yourself.

A few examples:

````python
my_list = [1, 2, 3, 5, 7]

for number in my_list:
    # one after another, we add 2 to each number in the list
    result = number + 2
    # and print the result
    print(number, "+ 2 =", result)

# back outside the code block
print("this only happens once")
````

````python
# if you want to do something x times

x = 10

for i in range(x):
    print(i)
````

````python
# combined with if-else

for number in [1, 2, 3, 4]:
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
````

````python
# if you want to iterate over a dictionary

d = {1: 123, 5: 22343, 12: 938493}
for key in d.keys(): # iterates over keys
    value = d[key]
    print(f"key: {key}, value: {value}") # this is a formatted string: you put an f before the first ", and then you can use variables in curly brackets
````

If you want to iterate over a list but also want to know the index, you can use `enumerate()`. This generates two items per iteration: The index of the list item, and the list item.

````python
my_list = [123, 234, 345]

for i, item in enumerate(my_list):
    print(f"list item: {item}, index: {i}")
````

You can also make multi-demensional `for`-loops:

````python
for i in range(10):
    for j in range(10):
        print(i * j, end="\t") # \t is the symbol for a tab

    print() # this is back in the outer for-loop and starts a new line
````

````python
# if you want to skip code, you can use the key-word "continue" to move to the next iteration

for i in range(10):
    print(i)          # <-
    if i == 5:        #   |
        print("skip") #   |
        continue      #  -     continue moves on to the next iteration, in this case: i = 6
    print("->", i*2)
````

````python
# if you want to stop the loop before it has looked at all elements, you can use the keyword "break"

for i in [123, 12344, 0, 2334]:
    if i == 0: 
        print("cannot divide by 0")
        break # the for loop is interrupted - the code below the loop is executed next
    print(12/i)

print("after the loop")
````

### `while`-loops (bonus)

`while`-loops are a bit more versatile but they are also error-prone. They do not iterate over an object - instead they just repeat the code until a condition isn't met anymore or they interupted by `break`.

The syntax is:

```py
while condition:
    # do something
```

The loop runs as long as `condition` is `True`. If it is set to `False`, the loop stops before the next iteration:

````python
condition = True
string = ""

while condition:
    # add an extra character to the string
    string += 'N' # this is the same as "string = string + 'N'" -> it modifies the variable

    # check if the string has reached maximum length
    if len(string) >= 6:
        # the string has reached maximum length, stop after this iteration
        condition = False

    # this is still executed one last time, after we set condition to false
    print(string)
````

````python
# or, simpler:
string = ""

while len(string) < 6:
    string += 'N'
    print(string)
````

````python
# or we start an infinite loop and break it instead of setting a condition to false
string = ""

while True:
    string += 'N'

    # check if the string has reached maximum length
    if len(string) >= 6:
        # the string has reached maximum length, stop right now
        break

    # this is not executed one last time, after we broke the loop
    print(string)
````

We can use the `while`-loop to perform the same function as a `for`-loop by using a counter:

````python
# iterate over a list
my_list = [1, 6, 2, 7, 1, 55, 2, 6, 2]
i = 0

while i < len(my_list):
    print(my_list[i]) # use your counter i to access each element in the list
    i += 1
````

````python
# instead of using "for i in range(x)"
i = 0
x = 10

while i < x:
    print(i)
    i += 1
````

This can be error-prone, though. You have to check that you increase your counter in the right place (usually the end of the loop block) and that your condition is set correctly, to avoid trying to access a list element that doesn't exist.

## 5. Writing Functions

We now come to the last theory block, where you will learn to write Python functions yourself. A little recapitulation from chapter 2: Functions take arguments, do something with them, and may return some data.

For example the function `sum()` returns the sum of the elements in the list given.

````python
my_list = [1 + 2 + 3]
sum(my_list)
````

````python
def my_sum_function(x):     # take arguments
    result = 0              # do something
    for element in x:       # ...
        result += element   # ...
    return result           # return a result
````

````python
my_sum_function(my_list)
````

When you define a new function you always start with `def`, a space, and then the name of the function (here `my_sum_funciton`) and the arguments in brackets. Here, we call the argument `x`, but you can choose the name. You can put as many arguments as you need, separated by commas. 

Then, indented by one tab, the body of the funtion follows. Here you can use your previously defined arguments as variables. If you want your function to return something, you can use the keyword `return`. It is commonly used at the end of a function, but can be used anywhere - but you have to be aware that after a `return`-statement is reached, the rest of the code in the function is not executed. You can use this to achieve different outcomes in an if/else-statement.

Some examples:

````python
def add(a, b):
    return a+b 

add(3, 4)
````

````python
# we don't have to return anything

def print_list_elements(some_list):
    for element in some_list:
        print(element)

print_list_elements([1, 2, 3, 4])
````

````python
# the variable "some_list" does not exist outside of the function body
print(some_list)
````

````python
# we can also modify lists and dictionaries in-place, without returning anything

def change_first_list_element_to_10k(a_list):
    if len(a_list) > 0:
        a_list[0] = 10000

l = [1, 2, 3]
change_first_list_element_to_10k(l)
print(l)
````

````python
# this does not work for more basic types such as integers

def double_number(number):
    number = number * 2

n = 12
double_number(n)
print(n)

# if you want to change an integer with a function, you have to return the new value and assign it to the variable

def double_number_v2(number):
    return number * 2

n = 12
n = double_number_v2(n)
print(n)
````

````python
# if you put a return statement in the middle of the function body, 
# everything after the return statement does not get executed

def print_some_lines(): # no arguments also works
    print("Lorem ipsum dolor sit amet")
    print("onsetetur sadipscing elitr")
    return 1
    print("sed diam nonumy eirmod") # this will not be printed

print_some_lines()
````

````python
# this can be useful -> we can add a conditional interruption of the function

def remove_last_list_item(a_list):
    if len(a_list) == 0: # check that the list isn't empty
        print("no items left to remove")
        return # you can also use an empty return statement to return noting (returns a "None" value)
        # but still terminate the function
    
    # we do not need an else statement now
    last_element = a_list.pop() # pop removes the last element of a list and gives it to us
    return last_element

my_list = [1, 2, 3]
for item in range(5):
    print(remove_last_list_item(my_list))
````

````python
# both of these functions do the same thing 

def goat_age_category(goat_age):
    category = ""
    if goat_age < 3: 
        category = "baby goat"
    elif goat_age > 12: 
        category = "senior goat"
    else:
        category = "adult goat"

    return category

def goat_age_category_v2(goat_age):
    if goat_age < 3: 
        return "baby goat"
    if goat_age > 12: 
        return "senior goat"
    return "adult goat"
````

````python
# we can also return multiple elements

def add_subtract_multiply_divide(a, b):
    res_add = a + b
    res_sub = a - b
    res_mul = a * b
    res_div = a / b

    return res_add, res_sub, res_mul, res_div

print(add_subtract_multiply_divide(3, 4))

# or assign to a new variable
````

Putting your code into functions has two main advantages: Simplyfing long scripts by putting separate steps into functions, and if you want to do something similar multiple times, with different values or variables.

````python
print(add_subtract_multiply_divide(3, 4))
print(add_subtract_multiply_divide(4, 4))
print(add_subtract_multiply_divide(5, 4))
print(add_subtract_multiply_divide(6, 5))
print(add_subtract_multiply_divide(2, 4))
print(add_subtract_multiply_divide(1, 4))
````
