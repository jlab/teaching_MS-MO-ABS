```python
# this should print each element, its length, and its reverse

fruits = [
    "apple",
    "banana",
    "peach",
    "pear",
    "apricot"
]

i = 0

while i < len(fruits):
    length = len(fruits[i])
    rev = "".join((list(reversed(fruits[i]))))
    i += 1
    print(f"fruit: {fruits[i]}, length: {length}, reverse: {rev}")
```
